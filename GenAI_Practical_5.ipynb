```python
from transformers import T5ForConditionalGeneration, T5Tokenizer

# Load pre-trained model and tokenizer
model_name = "t5-small"
tokenizer = T5Tokenizer.from_pretrained(model_name)
model = T5ForConditionalGeneration.from_pretrained(model_name)
```

### Step 3: Prepare the input data and generate summary

Provide a sample news article and use the loaded model to generate its abstractive summary.

```python
article = """\nLondon (CNN) -- King Charles III and Queen Camilla will be crowned at Westminster Abbey on Saturday in a ceremony steeped in ancient tradition but updated for a modern age. More than 2,200 guests will be in attendance for the first coronation of a British monarch in 70 years, and the first time a Queen Consort has been crowned in more than a century.\nThe event comes eight months after the death of Queen Elizabeth II, the longest-reigning monarch in British history, who died at the age of 96 after 70 years on the throne. Charles immediately became King upon his mother's death, but the coronation formally marks the start of his reign.\nWhile the pageantry and spectacle of the coronation will take place in London, celebrations will be held across the United Kingdom and around the world, with millions expected to watch the historic event.\nThe service will begin at 11 a.m. local time (6 a.m. ET) and is expected to last for around two hours, a shorter ceremony than that of his mother, which ran for three hours. The King and Queen will travel in the Gold State Coach, a more than 260-year-old carriage, in a procession from Buckingham Palace to Westminster Abbey.\n"""

# Prepare the input for the model
input_text = "summarize: " + article
input_ids = tokenizer.encode(input_text, return_tensors="pt", max_length=512, truncation=True)

# Generate the summary
summary_ids = model.generate(input_ids, max_length=150, min_length=40, length_penalty=2.0, num_beams=4, early_stopping=True)
summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

print("Original Article:\n", article)
print("\nGenerated Summary:\n", summary)
```
